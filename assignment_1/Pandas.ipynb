```python
#1. Create a pandas DataFrame from the CSV
import pandas as pd
df=pd.read_csv("sales_sample.csv")
print(df)
```

```python
#dataframe from lists
transaction_id=[1005,1006,1007]
store_id=[1,2,4]
date=['2025-10-11','2025-10-15','2025-10-12']
product_name=['Notebook','Stapler','Pen']
df_list=pd.DataFrame({
    "transaction_id":transaction_id,
    "store_id":store_id,
    "date":date,
    "product_name":product_name
})
print(df_list)
```

```python
#dataframe from dictionary
data={
    "product_id":[501,502,503],
    "product_name":["Notebook","Pen","Stapler"],
    "quantity":[5,3,4]
}
df_dict=pd.DataFrame(data)
print(df_dict)
```

```python
#2. Filter and subset rows/columns 
#Sales from a specific store(2)
store2_sales=df[df["store_id"]==2]
print(store2_sales.head())
```

```python
#Sales within a date range(10-12 october)
df["date"]=pd.to_datetime(df["date"])
date_filter=df[(df["date"]>="2025-10-10") & (df["date"]<="2025-10-12")]
print(date_filter)
```

```python
#high-value transactions
high_val=df[df["total"]>35]
print(high_val)
```

```python
promo_used = df[df["promo_code"].notna()] 
print(promo_used.head())
```

```python
subset_cols=df[["date","store_name","total"]]
print(subset_cols.head())
```

```python
#3.Compute descriptive statistics and group summaries (totals by store, average basket).
#average basket
import numpy as np
avg=np.mean(df["total"])
print(avg)
```

```python
df.describe()
```

```python
#total by store
total_store1=df.groupby("store_name")["total"].sum().reset_index()
print(total_store1)
```

```python
#Quantity sold by product
quant_prod=df.groupby("product_name")["quantity"].sum().reset_index()
print(quant_prod)
```

```python
#day wise sales
daily_sales=df.groupby("date")["total"].sum().reset_index()
print(daily_sales)
```

```python
#4.Perform simple cleaning (handle missing values, fix data types, drop duplicates).
df.isna().sum()
```

```python
#Filling missing values
df["promo_code"]=df["promo_code"].fillna("NONE")
```

```python
#Converting customer id to int
df["customer_id"]=df["customer_id"].astype(int)
```

```python
df["customer_id"]=df["customer_id"].fillna(df["customer_id"].mean())
```

```python
print(df)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
#Fixing datatypes
df["transaction_id"]=df["transaction_id"].astype(int)
df["store_id"]=df["store_id"].astype(int)
df["product_id"]=df["product_id"].astype(int)
```

```python
#Duplicates
df.duplicated().sum()
```

```python
#5.Build a mini ETL pipeline: read CSV → clean & transform → output JSON.
output_path = "sales.json"
df.to_json(output_path,orient="records",indent=4)
```
