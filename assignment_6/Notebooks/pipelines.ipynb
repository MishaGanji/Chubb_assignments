```python
%sql
CREATE OR REPLACE TABLE main.default.silver_sales_transactions
USING DELTA
AS
SELECT *
FROM delta.`/Volumes/main/default/datalake_retail/silver/sales_transactions`;
```

```python
%sql
CREATE OR REPLACE TABLE main.default.silver_products
USING DELTA
AS
SELECT *
FROM delta.`/Volumes/main/default/datalake_retail/silver/products`;
```

```python
%sql
CREATE OR REPLACE TABLE main.default.silver_stores
USING DELTA
AS
SELECT *
FROM delta.`/Volumes/main/default/datalake_retail/silver/stores`;
```

```python
%sql
CREATE OR REPLACE TABLE main.default.silver_quarantine
USING DELTA
AS
SELECT *
FROM delta.`/Volumes/main/default/datalake_retail/silver/sales_quarantine`;
```

```python
%sql
SELECT COUNT(*) FROM main.default.silver_sales_transactions;
SELECT COUNT(*) FROM main.default.silver_products;
SELECT COUNT(*) FROM main.default.silver_stores;
```

```python
%sql
CREATE OR REPLACE TABLE main.default.daily_sales_summary
USING DELTA
AS
SELECT
    transaction_date,
    SUM(total_amount) AS daily_revenue,
    SUM(quantity) AS total_quantity_sold,
    COUNT(transaction_id) AS transaction_count
FROM main.default.silver_sales_transactions
GROUP BY transaction_date;
```

```python
%sql
CREATE OR REPLACE TABLE main.default.monthly_revenue_region
USING DELTA
AS
SELECT
    year(s.transaction_date) AS year,
    month(s.transaction_date) AS month,
    st.region,
    SUM(s.total_amount) AS monthly_revenue
FROM main.default.silver_sales_transactions s
JOIN main.default.silver_stores st
    ON s.store_id = st.store_id
GROUP BY
    year(s.transaction_date),
    month(s.transaction_date),
    st.region;
```
