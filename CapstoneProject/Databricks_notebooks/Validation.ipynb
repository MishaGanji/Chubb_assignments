KPI VALIDATION EXECUTIVE SALES

```python
%sql
SELECT 
  SUM(total_quantity_sold) 
FROM misha_azure.default.gold_sales_performance;
```

```python
%sql
SELECT 
  SUM(total_revenue) AS total_revenue
FROM misha_azure.default.gold_sales_performance;
```

```python
%sql
SELECT 
  SUM(net_revenue)
FROM misha_azure.default.gold_region_performance;
```

KPI VALIDATION REGIONAL AND STORE PERFORMANCE

```python
%sql
SELECT 
  region,
  SUM(net_revenue) AS net_revenue
FROM misha_azure.default.gold_region_performance
GROUP BY region
ORDER BY net_revenue DESC;
```

KPI VALIDATION PRODUCT AND CUSTOMER ANALYTICS

```python
%sql
SELECT 
  COUNT(customer_name)
FROM misha_azure.default.gold_customer_analytics;
```

```python
%sql
SELECT 
  category,
  SUM(total_profit)
FROM misha_azure.default.gold_product_performance
GROUP BY category
ORDER BY SUM(total_profit) DESC;
```
